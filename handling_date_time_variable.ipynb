Working with Dates

```python
# Converting to datetime datatype
date['date'] = pd.to_datetime(date['date'])

date.info()
```

1. Extract year

```python
date['date_year'] = date['date'].dt.year

date.sample(5)
```

2. Extract Month

```python
date['date_month_no'] = date['date'].dt.month

date.head()
```

```python
date['date_month_name'] = date['date'].dt.month_name()

date.head()
```

Extract Days


```python
date['date_day'] = date['date'].dt.day

date.head()
```

```python
# day of week
date['date_dow'] = date['date'].dt.dayofweek

date.head()
```

```python
# day of week - name

date['date_dow_name'] = date['date'].dt.day_name()

date.drop(columns=['product_id','city_id','orders']).head()
```

```python
# is weekend?

date['date_is_weekend'] = np.where(date['date_dow_name'].isin(['Sunday', 'Saturday']), 1,0)

date.drop(columns=['product_id','city_id','orders']).head()
```

Extract week of the year

```python
# Ensure that the 'date' column is in datetime format
date['date'] = pd.to_datetime(date['date'])

# Now extract the week of the year using the isocalendar() method
date['date_week'] = date['date'].apply(lambda x: x.isocalendar()[1])

# Dropping unnecessary columns
date.drop(columns=['product_id', 'city_id', 'orders']).head()
```

Extract Quarter

```python
date['quarter'] = date['date'].dt.quarter

date.drop(columns=['product_id','city_id','orders']).head()
```

```python
# date['semester'] = np.where(date['quarter'].isin([1,2]), 1, 2):

# date['quarter']: This accesses the quarter column in the date DataFrame.
# isin([1,2]): This checks if the value in the quarter column is either 1 or 2.
#  It returns a boolean result (True if the value is in [1, 2], otherwise False).
# np.where(condition, value_if_true, value_if_false): This is a conditional function from numpy.
# If the condition is True (i.e., if the value in quarter is 1 or 2),
# then 1 is assigned to the corresponding row in the new semester column.
# If the condition is False (i.e., if the value in quarter is neither 1 nor 2), then 2 is assigned to the semester column.
# Result: The semester column will be filled with 1 for records where quarter is 1 or 2, and 2 for records where quarter is 3 or 4.

# Example:

# If a row has quarter = 1, the new column semester will have 1.
# If a row has quarter = 3, the new column semester will have 2.
date['semester'] = np.where(date['quarter'].isin([1,2]), 1, 2)

date.drop(columns=['product_id','city_id','orders']).head()
```

Extract Time elapsed between dates

```python
import datetime
#The datetime.today() method from the datetime class returns the current local date and time
#  (i.e., the "now" moment when the code is run).
today = datetime.datetime.today()

today
```

```python
#Finally, it calculates the difference between the current date (today) and the date from a column (date['date']) in a DataFrame,
# yielding a time difference (timedelta).
today - date['date']
```

```python
(today - date['date']).dt.days
```
