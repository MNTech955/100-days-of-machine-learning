```python
from sklearn.datasets import make_regression

import numpy as np
import matplotlib.pyplot as plt
```

```python
%matplotlib notebook
from matplotlib.animation import FuncAnimation
import matplotlib.animation as animation
```

```python
X,y = make_regression(n_samples=100, n_features=1, n_informative=1, n_targets=1,noise=20,random_state=13)
```

```python
plt.scatter(X,y)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

# Convert X to 1D to avoid shape problems
X_flat = X.ravel()

# Gradient Descent values
b = -520
m = 600
lr = 0.001

all_b = []
all_m = []
all_cost = []

epochs = 30

for i in range(epochs):
    slope_b = 0
    slope_m = 0
    cost = 0

    for j in range(X_flat.shape[0]):
        y_pred = m * X_flat[j] + b
        error = y[j] - y_pred

        slope_b = slope_b - 2 * error
        slope_m = slope_m - 2 * error * X_flat[j]
        cost = cost + error ** 2

    b = b - (lr * slope_b)
    m = m - (lr * slope_m)

    all_b.append(b)
    all_m.append(m)
    all_cost.append(cost)
```

```python
fig, ax = plt.subplots(figsize=(9, 5))

x_i = np.arange(-3, 3, 0.1)

ax.scatter(X_flat, y)

line, = ax.plot(x_i, x_i * all_m[0] + all_b[0], 'r-', linewidth=2)

ax.set_xlim(-3, 3)
ax.set_ylim(min(y) - 50, max(y) + 50)

def update(i):
    label = 'epoch {}'.format(i + 1)
    line.set_ydata(x_i * all_m[i] + all_b[i])
    ax.set_xlabel(label)
    return line,

anim = FuncAnimation(
    fig,
    update,
    frames=epochs,
    interval=500,
    repeat=True
)



HTML(anim.to_jshtml())
```
